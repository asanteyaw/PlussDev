## PlussMa DataFrame with HTML Rendering

To get the same rendered HTML output with PlussMa DataFrames, use the `to_html()` method with `IPython.display.HTML()`:

```python
# Create a C++ program that outputs HTML
cpp_code = '''
#include <pluss/datalog/table/dataframe/dataframe.h>
#include <iostream>

int main() {
    auto df = pluss::table::DataFrame::load("csv", "../data/messy_data.csv");
    std::cout << df->to_html();
    return 0;
}
'''

# In practice, you would use Python bindings
# For demonstration, we'll show how to render HTML from the C++ output
html_output = """<!-- HTML output from df->to_html() would go here -->"""

print("PlussMa DataFrame rendered as HTML:")
display(HTML(html_output))
```

## Python Wrapper Example

Here's how you would create a Python wrapper for PlussMa DataFrames that automatically renders HTML in notebooks:

```python
class DataFrameWrapper:
    """Python wrapper for PlussMa DataFrame with Jupyter notebook support"""
    
    def __init__(self, cpp_dataframe):
        self.df = cpp_dataframe
    
    def head(self, n=5):
        """Display first n rows with HTML rendering in notebooks"""
        html = self.df.to_html()  # Call C++ to_html() method
        return HTML(html)
    
    def tail(self, n=5):
        """Display last n rows with HTML rendering in notebooks"""
        html = self.df.to_html()  # Call C++ to_html() method  
        return HTML(html)
    
    def describe(self):
        """Statistical summary with HTML rendering"""
        # Call C++ describe method that returns HTML
        html = self.df.describe_html()
        return HTML(html)
    
    def _repr_html_(self):
        """Automatic HTML rendering when displayed in notebook"""
        return self.df.to_html()

# Example usage:
# df = DataFrameWrapper(cpp_df)
# df.head(3)  # Automatically renders as HTML table
```

## Summary

**Key Points:**

1. **Terminal/Console**: Use default methods like `df->head()` which output plain text
2. **Jupyter Notebooks**: Use `df->to_html()` and wrap with `IPython.display.HTML()`
3. **Python Bindings**: Create a wrapper class with `_repr_html_()` method for automatic rendering
4. **LaTeX Output**: Use `df->to_latex()` for LaTeX documents

The HTML output uses the same CSS classes as pandas (`dataframe` class), so you can style it with custom CSS.
